# Data Cleaning

Data cleaning helps ensure that the analysis is accurate and reliable.

The dataset will be checked for:

```python
# Check missing values

print("Missing Values:")
print(df.isnull().sum())
```

# Data Analysis

The following analysis is performed to answer the questions provided in the task.

## Question 1: Average Final Grade (G3)

The average final grade is calculated to understand the overall academic performance of students.

```python
average_grade = df["G3"].mean()

print("Average Final Grade (G3):")
print(round(average_grade, 2))
```

### Observation

The average final grade (G3) is 10.42.

The final grades in the dataset range from 0 to 20, and the average score of 10.42 lies near the middle of this range.

## Question 2: Students Scoring Above 15

This analysis identifies the number of students who achieved a final grade greater than 15.

```python
high_performers = df[df["G3"] > 15]

print("Number of Students Scoring Above 15:")
print(len(high_performers))
```

### Observation

A total of 40 students scored above 15 in the final grade.

This shows that only a small proportion of students achieved high academic performance.

## Question 3: Relationship Between Study Time and Final Grade

The correlation between study time and final grade is calculated to determine whether students who study more tend to achieve higher grades.

```python
correlation = df["studytime"].corr(df["G3"])

print("Correlation Between Study Time and Final Grade:")
print(round(correlation, 3))
```

### Observation

The correlation between study time and final grade is 0.098.

This indicates a very weak positive relationship between study time and academic performance. Based on this dataset, study time alone does not appear to have a strong influence on final grades.

## Question 4: Average Performance by Gender

This analysis compares the average final grades of male and female students.

```python
gender_average = df.groupby("sex")["G3"].mean()

print("Average Final Grade by Gender:")
print(gender_average)
```

### Observation

The average final grade for female students is 9.97, while the average final grade for male students is 10.91.

Male students performed slightly better on average than female students in this dataset.

# Data Visualization

Visualizations help present the findings in a clear and understandable way.

## Histogram of Final Grades

The histogram shows how grades are distributed among students.

```python
plt.figure(figsize=(8,4))

plt.hist(df["G3"], bins=10, edgecolor="black")

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

### Observation

The histogram shows that most students scored between 8 and 15 in their final grade (G3).

The highest concentration of students is around grades 10 to 12, indicating that the majority achieved moderate academic performance. Very few students obtained extremely high grades (above 18) or very low grades.

Overall, the distribution suggests that student performance is centered around the middle grade range.

## Scatter Plot: Study Time vs Final Grade

The scatter plot is used to visualize the relationship between study time and final grades.

```python
plt.figure(figsize=(8,4))

plt.scatter(df["studytime"], df["G3"])

plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Final Grade (G3)")

plt.show()
```

### Observation

The scatter plot shows a wide spread of grades across all study time categories.

Although students with higher study time also achieve some higher grades, there is no clear upward trend in the data. This supports the correlation value of 0.098, which indicates a very weak positive relationship between study time and final grade.

## Bar Chart: Average Grade by Gender

This chart compares the average final grades of male and female students.

```python
gender_average.index = ["Female", "Male"]
gender_average.plot(kind="bar", figsize=(6,5), edgecolor= "black")

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Grade")

plt.xticks(rotation=0)

plt.show()
```

### Observation

The bar chart shows that male students have a slightly higher average final grade than female students.

The average grade for male students is 10.91, while the average grade for female students is 9.97. However, the difference between the two groups is relatively small.
